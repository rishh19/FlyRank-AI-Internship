```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset loaded successfully.")
print("Total pages:", len(df))
```

##2. The Question: Decision, Action, and Cost

**Research Question:**
Which web pages should be prioritized for content refresh based on observable performance signals such as impressions, CTR, content age, and traffic trends?

**Decision:**
This project aims to improve the decision of which pages should be reviewed first for content updates.

**Action:**
The SEO or content team can use the ranked recommendations to prioritize their limited time and focus on pages that are most likely to benefit from a refresh.

**Cost of a Wrong Recommendation:**
If the recommendations are incorrect, the team may spend valuable time reviewing pages that do not need updates, while genuinely declining pages may be missed. This can reduce the effectiveness of content maintenance and delay improvements to important pages.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
priority_pages = df[df["trend_direction"] == "down"]

print("Pages that would be prioritized for review:", len(priority_pages))
print("Percentage of dataset:",
      round(len(priority_pages) / len(df) * 100, 2), "%")
```

## 3.Quick Look at the Data

The starter dataset contains **30,000** web pages, of which **16,262** are currently marked as declining. The average content age is **256.17 days**. These numbers show that a large number of pages may require review, making it useful to build a machine learning system that helps prioritize which pages should be refreshed first.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
avg_content_age = df["content_age_days"].mean()

print(f"Total pages: {total_pages}")
print(f"Declining pages: {declining_pages}")
print(f"Average content age: {avg_content_age:.2f} days")
```
